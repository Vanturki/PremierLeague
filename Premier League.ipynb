```python
top_scorers = players[['Player', 'Squad', 'Performance_Gls']].sort_values(
    'Performance_Gls', ascending=False).head(10)

print(top_scorers)
```

```python
plt.figure(figsize=(12, 6))
plt.barh(top_scorers['Player'], top_scorers['Performance_Gls'], color='steelblue')
plt.xlabel('HOW MANY GOALS')
plt.title('TOP SCORERS 2024-2025')
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

top_scorers = players[['Player', 'Squad', 'Performance_Gls']].sort_values(
    'Performance_Gls', ascending=False).head(10)

fig, ax = plt.subplots(figsize=(12, 7))
bars = ax.barh(top_scorers['Player'], top_scorers['Performance_Gls'], 
               color='steelblue', edgecolor='white')

# إضافة الأرقام على كل بار
for bar, val in zip(bars, top_scorers['Performance_Gls']):
    ax.text(bar.get_width() + 0.2, bar.get_y() + bar.get_height()/2,
            f'{int(val)} goals', va='center', fontsize=11, color='black')

ax.set_xlabel('Number of Goals', fontsize=12)
ax.set_title('Top 10 Scorers - Premier League 2024/25 ⚽', fontsize=14, fontweight='bold')
ax.invert_yaxis()
plt.tight_layout()
plt.show()
```

```python
# ensure necessary modules are available (useful if the kernel was restarted)
if 'pd' not in globals():
    import pandas as pd
if 'np' not in globals():
    import numpy as np
if 'plt' not in globals():
    import matplotlib.pyplot as plt

# ensure the dataframe is loaded (useful if the kernel was restarted)
if 'players' not in globals():
    from pathlib import Path
    players = pd.read_csv(Path("data") / "Squad_PlayerStats__stats_standard.csv")

top10 = players[['Player', 'Performance_Gls', 'Performance_Ast']].sort_values(
        'Performance_Gls', ascending=False).head(10)

x = np.arange(len(top10['Player']))
width = 0.35

fig, ax = plt.subplots(figsize=(14, 7))

bars1 = ax.bar(x - width/2, top10['Performance_Gls'], width, 
               label='Goals ⚽', color='steelblue')
bars2 = ax.bar(x + width/2, top10['Performance_Ast'], width, 
               label='Assists 🎯', color='orange')

# إضافة الأرقام فوق كل عمود
for bar in bars1:
    ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.1,
            str(int(bar.get_height())), ha='center', fontsize=9)

for bar in bars2:
    ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.1,
            str(int(bar.get_height())), ha='center', fontsize=9)

ax.set_xticks(x)
ax.set_xticklabels(top10['Player'], rotation=45, ha='right')
ax.set_title('Goals vs Assists - Top 10 Players ⚽🎯', fontsize=14, fontweight='bold')
ax.legend()
plt.tight_layout()
plt.show()
```

```python
# أكثر الفرق في البطاقات
cards = players.groupby('Squad')[['Performance_CrdY', 'Performance_CrdR']].sum()
cards = cards.sort_values('Performance_CrdY', ascending=False)

cards.plot(kind='bar', figsize=(14, 6), color=['gold', 'red'])
plt.title('Yellow & Red Cards by Team', fontsize=14, fontweight='bold')
plt.xlabel('Team')
plt.ylabel('Number of Cards')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
